```python
monate = ["01/2025", "02/2025", "03/2025", "04/2025", "05/2025", "06/2025"]
ist = [14500, 13900, 12800, 11600, 12300, 13500]
prognose = [14300, 14100, 12600, 11900, 12500, 13200]
fig = eda.timeseries_forecast(monate, ist, prognose, title="Demodaten · Januar–Juni 2025", y_title="Verbrauch (kWh)")
fig.show(renderer="notebook")  # Plotly wird offline eingebettet.
```

```python
residuen = [actual - predicted for actual, predicted in zip(ist, prognose)]
eda.residual_bars(monate, residuen, schwelle=250, title="Demodaten · Residuen Januar–Juni 2025").show(renderer="notebook")
```
